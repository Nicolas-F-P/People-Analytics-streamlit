## 2. Nettoyage initial des données

```python
print(f"Nombre de lignes :{df.shape[0]}")
print(f"Nombre de colonnes : {df.shape[1]}")
```

```python
missing_values = df.isna().sum().sort_values(ascending=False)
missing_values
```

```python
duplicates_count = df.duplicated().sum()
duplicates_count
```

```python
unique_values = df.nunique().sort_values()
unique_values
```

```python
constant_columns = [col for col in df.columns if df[col].nunique() == 1]
constant_columns
```

```python
df_clean = df.copy()
df_clean.head()
```

```python
df_clean["Attrition_Flag"] = df_clean["Attrition"].map({"Yes": 1, "No": 0})
df_clean[["Attrition", "Attrition_Flag"]].head()
```

```python
columns_to_drop = ["EmployeeCount", "Over18", "StandardHours", "EmployeeNumber"]
df_clean = df_clean.drop(columns=columns_to_drop)
df_clean.head()
```

```python
print(f"Ancienne taille : {df.shape}")
print(f"Nouvelle taille : {df_clean.shape}")
```

```python
attrition_count = df_clean["Attrition"].value_counts()
attrition_count
```

```python
attrition_rate = df_clean["Attrition_Flag"].mean() * 100
print(f"Taux d'attrition : {attrition_rate:.2f}%")
```

```python
output_path = Path("../data/processed/employee_attrition_clean.csv")
output_path.parent.mkdir(parents=True, exist_ok=True)
df_clean.to_csv(output_path, index=False)
print("Dataset nettoyé sauvegardé avec succès.")
```
